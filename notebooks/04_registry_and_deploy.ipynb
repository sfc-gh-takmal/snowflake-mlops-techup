# 04 - Model Registry & Inference Deployment

This notebook covers:
1. Registering a trained model to the Snowflake Model Registry
2. Running batch inference via the Model Registry
3. Deploying a real-time inference service (SPCS) with Gateway
4. Testing predictions via SQL and Python

**Key Concepts:**
- **Model Registry**: Versioned model storage with dependency tracking
- **Batch Inference**: `model_version.run()` on a warehouse
- **SPCS Inference**: Containerized model serving on compute pools
- **Gateway**: HTTP endpoint for external REST API access

```python
import sys
sys.path.insert(0, "../source")
from snowpark_session import create_snowpark_session
from config import DATABASE, SCHEMA, WAREHOUSE, MODEL_NAME

session = create_snowpark_session()
session.sql(f"USE WAREHOUSE {WAREHOUSE}").collect()
print(f"Connected as: {session.get_current_user()}")
```

## Step 1: Register Model to Snowflake Model Registry

Assumes you ran notebook 03 and have `/tmp/model.ubj` and `/tmp/sample_input.json` saved.

```python
import pandas as pd
import xgboost as xgb
from snowflake.ml.registry import Registry

# Load model artifact
model = xgb.XGBClassifier()
model.load_model("/tmp/model.ubj")
sample_input = pd.read_json("/tmp/sample_input.json", orient="records")

# Initialize registry
reg = Registry(session=session, database_name=DATABASE, schema_name=SCHEMA)

# Auto-increment version
try:
    versions_df = session.sql(f"SHOW VERSIONS IN MODEL {DATABASE}.{SCHEMA}.{MODEL_NAME}").collect()
    existing = [r["name"] for r in versions_df]
    max_v = max(int(v.replace("V", "")) for v in existing if v.startswith("V") and v[1:].isdigit())
    version_name = f"V{max_v + 1}"
except Exception:
    version_name = "V1"

print(f"Registering as: {DATABASE}.{SCHEMA}.{MODEL_NAME}/{version_name}")
```

```python
# Register the model
mv = reg.log_model(
    model=model,
    model_name=MODEL_NAME,
    version_name=version_name,
    conda_dependencies=["xgboost==3.3.0", "scikit-learn"],
    sample_input_data=sample_input,
    comment=f"XGBoost fraud detector - notebook registration",
)

print(f"Registered: {MODEL_NAME}/{version_name}")
print(f"\nModel info:")
reg.show_models()
```

## Step 2: Batch Inference (Warehouse)

Use the Model Registry's `run()` method to score data on a warehouse. No service deployment needed.

```python
# Get the model version
m = reg.get_model(MODEL_NAME)
mv = m.version(version_name)

# Run batch inference on sample data
test_df = session.create_dataframe(sample_input)
predictions = mv.run(test_df, function_name="predict_proba")
predictions.show(5)

print(f"\nScored {predictions.count()} rows via warehouse batch inference.")
```

## Step 3: Deploy Real-Time Service (SPCS)

Deploy the model as a containerized service for real-time predictions.

**Note:** This creates an SPCS service which takes 2-5 minutes to become ready (container image build + startup).

```python
from config import COMPUTE_POOL, SERVICE_NAME

# Deploy inference service
mv.create_service(
    service_name=f"{DATABASE}.{SCHEMA}.{SERVICE_NAME}",
    service_compute_pool=COMPUTE_POOL,
    image_build_compute_pool=COMPUTE_POOL,
    ingress_enabled=True,
    max_instances=1,
    force_rebuild=True,
)

print(f"Service deploying: {SERVICE_NAME}")
print("Wait 2-5 minutes for the service to become READY.")
```

## Step 4: Check Service Status

Re-run this cell until the service shows READY.

```python
import json as _json

status = session.sql(f"""
    SELECT SYSTEM$GET_SERVICE_STATUS('{DATABASE}.{SCHEMA}.{SERVICE_NAME}')
""").collect()

status_data = _json.loads(status[0][0])
for svc in status_data:
    print(f"  Status: {svc['status']}")
    print(f"  Message: {svc.get('message', '')}")
```

## Step 5: Test Inference via SQL

SQL-based inference works immediately via warehouse (no service needed):

```python
from config import FEATURE_COLUMNS, SCORING_VIEW_NAME

# Pass the feature columns explicitly. PREDICT_PROBA(*) over SELECT * would also
# feed TXN_ID, CUSTOMER_ID and FEATURE_TS into the model, which does not match the
# signature it was logged with and raises a signature mismatch.
feature_args = ", ".join(FEATURE_COLUMNS)

session.sql(f"""
    WITH sample AS (
        SELECT TXN_ID, {feature_args}
        FROM {DATABASE}.{SCHEMA}.{SCORING_VIEW_NAME}
        LIMIT 5
    )
    SELECT
        TXN_ID,
        {DATABASE}.{SCHEMA}.{MODEL_NAME}!PREDICT_PROBA({feature_args}) AS fraud_probability
    FROM sample
""").show()
```
