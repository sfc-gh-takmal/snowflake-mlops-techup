# 03 - Model Training

This notebook covers:
1. Loading training data from the Feature Store
2. Training an XGBoost classifier with cross-validation
3. Evaluating against quality gate thresholds
4. Saving model artifacts for registration

**Training approach:**
- XGBoost with `scale_pos_weight` to handle class imbalance (~3% fraud)
- 5-fold stratified cross-validation for robust evaluation
- Key metrics: AUC-ROC, PR-AUC, Precision, Recall, F1

**Note:** In the automated CI/CD pipeline, training runs as an ML Job on a compute pool
via the Task DAG (`source/pipeline/ml_pipeline_dag.py`). This notebook demonstrates
the same training logic interactively.

```python
import sys
sys.path.insert(0, "../source")
from snowpark_session import create_snowpark_session
from config import (
    DATABASE, SCHEMA, WAREHOUSE, SOURCE_DATABASE, SOURCE_SCHEMA,
    SCORING_VIEW_NAME, FEATURE_COLUMNS,
    N_ESTIMATORS, LEARNING_RATE, MAX_DEPTH, SCALE_POS_WEIGHT,
)

session = create_snowpark_session()
session.sql(f"USE WAREHOUSE {WAREHOUSE}").collect()
print(f"Connected as: {session.get_current_user()}")
print(f"Using warehouse: {WAREHOUSE}")
print(f"Features ({len(FEATURE_COLUMNS)}): {', '.join(FEATURE_COLUMNS)}")
```

## Step 1: Load Training Data from Feature Store

We join the Feature View (customer risk features) with the raw transactions table
to get labeled training data.

```python
import pandas as pd

# One row per transaction. The label is joined on TXN_ID, not CUSTOMER_ID.
#
# Joining customer-level features straight onto RAW_TRANSACTIONS (the original
# approach) produced ~20 rows per customer sharing one feature vector but carrying
# different labels -- 2,271 of 5,000 customers had both fraud and non-fraud rows.
# The target was unlearnable and precision was pinned to the 3% base rate.
cols = ", ".join(f"f.{c}" for c in FEATURE_COLUMNS)

df = session.sql(f"""
    SELECT
        f.TXN_ID,
        {cols},
        r.IS_FRAUD
    FROM {DATABASE}.{SCHEMA}.{SCORING_VIEW_NAME} f
    JOIN {SOURCE_DATABASE}.{SOURCE_SCHEMA}.RAW_TRANSACTIONS r ON f.TXN_ID = r.TXN_ID
""").to_pandas()

assert not df["TXN_ID"].duplicated().any(), "Feature join is fanning out -- duplicate TXN_IDs"

print(f"Dataset shape: {df.shape}")
print(f"Rows per transaction: {len(df) / df['TXN_ID'].nunique():.1f}")
print(f"Fraud rate: {df['IS_FRAUD'].mean():.2%}")
print("\nLabel distribution:")
print(df["IS_FRAUD"].value_counts())
```

## Step 2: Train XGBoost with Cross-Validation

```python
import numpy as np
import xgboost as xgb
from sklearn.model_selection import StratifiedKFold, train_test_split
from sklearn.metrics import roc_auc_score, average_precision_score, precision_score, recall_score, f1_score

# Feature contract and hyperparameters both come from config.py, so this notebook
# cannot drift from what the CI pipeline trains.
FEATURE_COLS = list(FEATURE_COLUMNS)

X = df[FEATURE_COLS].fillna(0)
y = df["IS_FRAUD"].astype(int)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

params = {
    "n_estimators": N_ESTIMATORS,
    "learning_rate": LEARNING_RATE,
    "max_depth": MAX_DEPTH,
    "scale_pos_weight": SCALE_POS_WEIGHT,
    "objective": "binary:logistic",
    "eval_metric": "aucpr",
    "random_state": 42,
}

# Cross-validation
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = []
for fold, (train_idx, val_idx) in enumerate(cv.split(X_train, y_train)):
    fold_model = xgb.XGBClassifier(**params)
    fold_model.fit(X_train.iloc[train_idx], y_train.iloc[train_idx], verbose=False)
    fold_proba = fold_model.predict_proba(X_train.iloc[val_idx])[:, 1]
    score = roc_auc_score(y_train.iloc[val_idx], fold_proba)
    cv_scores.append(score)
    print(f"  Fold {fold+1}: AUC-ROC = {score:.4f}")

print(f"\nCV Mean AUC-ROC: {np.mean(cv_scores):.4f} (+/- {np.std(cv_scores):.4f})")
```

## Step 3: Final Model Training & Evaluation

```python
# Train final model on full training set
model = xgb.XGBClassifier(**params)
model.fit(X_train, y_train, eval_set=[(X_test, y_test)], verbose=False)

# Evaluate on holdout test set
y_proba = model.predict_proba(X_test)[:, 1]
y_pred = (y_proba >= 0.5).astype(int)

metrics = {
    "auc_roc": roc_auc_score(y_test, y_proba),
    "pr_auc": average_precision_score(y_test, y_proba),
    "precision": precision_score(y_test, y_pred, zero_division=0),
    "recall": recall_score(y_test, y_pred, zero_division=0),
    "f1": f1_score(y_test, y_pred, zero_division=0),
    "cv_auc_mean": np.mean(cv_scores),
}

print("=== Model Metrics ===")
for k, v in metrics.items():
    print(f"  {k:15s}: {v:.4f}")
```

## Step 4: Quality Gate Check

```python
from config import MIN_AUC_ROC, MIN_PRECISION, MIN_RECALL

print("=== Quality Gate ===")
print(f"  AUC-ROC:   {metrics['auc_roc']:.4f} >= {MIN_AUC_ROC} -> {'PASS' if metrics['auc_roc'] >= MIN_AUC_ROC else 'FAIL'}")
print(f"  Precision: {metrics['precision']:.4f} >= {MIN_PRECISION} -> {'PASS' if metrics['precision'] >= MIN_PRECISION else 'FAIL'}")
print(f"  Recall:    {metrics['recall']:.4f} >= {MIN_RECALL} -> {'PASS' if metrics['recall'] >= MIN_RECALL else 'FAIL'}")

passed = (metrics["auc_roc"] >= MIN_AUC_ROC and
          metrics["precision"] >= MIN_PRECISION and
          metrics["recall"] >= MIN_RECALL)
print(f"
Overall: {'PASSED - ready for registration' if passed else 'FAILED - model not promoted'}")
```

## Step 5: Feature Importance

```python
importance = dict(zip(FEATURE_COLS, model.feature_importances_))
sorted_imp = sorted(importance.items(), key=lambda x: x[1], reverse=True)

print("=== Top Features ===")
for feat, imp in sorted_imp:
    bar = "#" * int(imp * 50)
    print(f"  {feat:30s} {imp:.4f} {bar}")
```

## Step 6: Save Model Artifact

Save the model locally. In the automated pipeline, this is saved to a Snowflake stage
and registered by the quality gate script.

```python
model.save_model("/tmp/model.ubj")
X_test.head(10).to_json("/tmp/sample_input.json", orient="records")
print("Model saved to /tmp/model.ubj")
print("Sample input saved to /tmp/sample_input.json")
print("
Next: Run notebook 04 to register this model in the Snowflake Model Registry.")

session.close()
```
